## 1. Load the CSV
Everything is read as text (`dtype=str`) so we can detect the errors ourselves.

```python
df = pd.read_csv('../data/sample_sales.csv', dtype=str, keep_default_na=False)
print(f'Rows read: {len(df)}')
df
```

## 2. Clean the data

```python
clean, discarded = clean_sales(df)
print(f'Valid rows: {len(clean)} | Discarded rows: {len(discarded)}')
```

## 3. Metrics

```python
metrics = calculate_metrics(clean)
print(f"Net sales: {metrics['net_sales']:,.2f} €")
print(f"Orders: {metrics['orders']}")
print(f"Average order value: {metrics['avg_order_value']:,.2f} €")
```
